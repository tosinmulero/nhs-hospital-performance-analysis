```python
import pandas as pd
from pathlib import Path

# Folder locations
raw_folder = Path("../data/raw")
processed_folder = Path("../data/processed")
processed_folder.mkdir(parents=True, exist_ok=True)

# Clean column names for the NHS Provider Level Data sheet
column_names = [
    "blank_start",
    "provider_code",
    "region",
    "provider_name",
    "type1_attendances",
    "type2_attendances",
    "type3_attendances",
    "total_attendances",
    "type1_under_4h",
    "type2_under_4h",
    "type3_under_4h",
    "total_under_4h",
    "type1_over_4h",
    "type2_over_4h",
    "type3_over_4h",
    "total_over_4h",
    "pct_within_4h_all",
    "pct_within_4h_type1",
    "pct_within_4h_type2",
    "pct_within_4h_type3",
    "emergency_admissions_type1",
    "emergency_admissions_type2",
    "emergency_admissions_type3_4",
    "total_emergency_admissions_via_ae",
    "other_emergency_admissions",
    "total_emergency_admissions",
    "decision_to_admit_over_4h",
    "decision_to_admit_over_12h",
    "blank_end"
]

# Month names in the filenames
month_map = {
    "April": "2026-04-01",
    "May": "2026-05-01",
    "June": "2026-06-01",
    "July": "2026-07-01",
    "August": "2026-08-01"
}

all_months = []

# Read all five NHS Excel files
for file in sorted(raw_folder.glob("*.xls")):

    print(f"Processing: {file.name}")

    raw = pd.read_excel(
        file,
        sheet_name="Provider Level Data",
        header=None
    )

    # Check workbook structure
    if raw.shape[1] != 29:
        raise ValueError(
            f"{file.name} has {raw.shape[1]} columns instead of 29"
        )

    # Provider records begin at Excel row 19
    monthly = raw.iloc[18:].copy()

    monthly.columns = column_names

    # Remove blank Excel formatting columns
    monthly = monthly.drop(
        columns=["blank_start", "blank_end"]
    )

    # Remove blank rows underneath the provider table
    monthly = monthly[
        monthly["provider_code"].notna()
    ].copy()

    # Clean text columns
    monthly["provider_code"] = (
        monthly["provider_code"]
        .astype(str)
        .str.strip()
    )

    monthly["region"] = (
        monthly["region"]
        .astype(str)
        .str.strip()
    )

    monthly["provider_name"] = (
        monthly["provider_name"]
        .astype(str)
        .str.strip()
    )

    # Identify reporting month from filename
    reporting_month = None

    for month_name, month_date in month_map.items():
        if month_name.lower() in file.name.lower():
            reporting_month = pd.Timestamp(month_date)
            break

    if reporting_month is None:
        raise ValueError(
            f"Could not identify reporting month for {file.name}"
        )

    monthly["reporting_month"] = reporting_month

    all_months.append(monthly)

    print(
        f"  Providers: {len(monthly)} | "
        f"Month: {reporting_month.strftime('%Y-%m')}"
    )

# Combine all five months
nhs_ae = pd.concat(
    all_months,
    ignore_index=True
)

# Convert analytical columns to numeric
text_columns = [
    "provider_code",
    "region",
    "provider_name",
    "reporting_month"
]

numeric_columns = [
    col for col in nhs_ae.columns
    if col not in text_columns
]

for col in numeric_columns:
    nhs_ae[col] = pd.to_numeric(
        nhs_ae[col],
        errors="coerce"
    )

# Sort chronologically and by provider
nhs_ae = nhs_ae.sort_values(
    ["reporting_month", "provider_code"]
).reset_index(drop=True)

# Save processed dataset
output_file = (
    processed_folder /
    "nhs_ae_provider_apr_aug_2026_clean.csv"
)

nhs_ae.to_csv(
    output_file,
    index=False
)

# Validation output
print("\n--- COMPLETE ---")
print("Combined shape:", nhs_ae.shape)

print("\nRows by month:")
print(
    nhs_ae.groupby("reporting_month")
    .size()
)

print("\nUnique providers:")
print(
    nhs_ae["provider_code"].nunique()
)

print("\nSaved to:")
print(output_file.resolve())
```

```python
# ============================================
# NHS A&E DATA QUALITY CHECKS
# ============================================

print("DATASET SHAPE")
print(nhs_ae.shape)

print("\nDATE RANGE")
print("Start:", nhs_ae["reporting_month"].min())
print("End:", nhs_ae["reporting_month"].max())

print("\nROWS BY MONTH")
print(
    nhs_ae.groupby("reporting_month")
    .size()
)

print("\nUNIQUE PROVIDERS")
print(nhs_ae["provider_code"].nunique())

# --------------------------------------------
# 1. Missing identifiers
# --------------------------------------------

print("\nMISSING IDENTIFIERS")

print(
    "Missing provider codes:",
    nhs_ae["provider_code"].isna().sum()
)

print(
    "Missing provider names:",
    nhs_ae["provider_name"].isna().sum()
)

print(
    "Missing regions:",
    nhs_ae["region"].isna().sum()
)

print(
    "Missing reporting months:",
    nhs_ae["reporting_month"].isna().sum()
)

# --------------------------------------------
# 2. Duplicate provider-month records
# --------------------------------------------

duplicate_provider_month = (
    nhs_ae.duplicated(
        subset=["provider_code", "reporting_month"],
        keep=False
    )
)

print("\nDUPLICATE PROVIDER-MONTH RECORDS")
print(duplicate_provider_month.sum())

# --------------------------------------------
# 3. Missing values by column
# --------------------------------------------

print("\nMISSING VALUES BY COLUMN")

missing_values = (
    nhs_ae.isna()
    .sum()
    .sort_values(ascending=False)
)

print(missing_values)

# --------------------------------------------
# 4. Negative numerical values
# --------------------------------------------

numeric_cols = nhs_ae.select_dtypes(
    include="number"
).columns

negative_counts = (
    nhs_ae[numeric_cols] < 0
).sum()

print("\nNEGATIVE VALUES")

print(
    negative_counts[
        negative_counts > 0
    ]
)

# --------------------------------------------
# 5. Attendance reconciliation
# total should equal type1 + type2 + type3
# --------------------------------------------

attendance_difference = (
    nhs_ae["total_attendances"]
    - (
        nhs_ae["type1_attendances"]
        + nhs_ae["type2_attendances"]
        + nhs_ae["type3_attendances"]
    )
)

print("\nATTENDANCE TOTAL MISMATCHES")
print(
    attendance_difference
    .fillna(0)
    .ne(0)
    .sum()
)

# --------------------------------------------
# 6. Under-4-hour reconciliation
# --------------------------------------------

under4_difference = (
    nhs_ae["total_under_4h"]
    - (
        nhs_ae["type1_under_4h"]
        + nhs_ae["type2_under_4h"]
        + nhs_ae["type3_under_4h"]
    )
)

print("\nUNDER-4-HOUR TOTAL MISMATCHES")
print(
    under4_difference
    .fillna(0)
    .ne(0)
    .sum()
)

# --------------------------------------------
# 7. Over-4-hour reconciliation
# --------------------------------------------

over4_difference = (
    nhs_ae["total_over_4h"]
    - (
        nhs_ae["type1_over_4h"]
        + nhs_ae["type2_over_4h"]
        + nhs_ae["type3_over_4h"]
    )
)

print("\nOVER-4-HOUR TOTAL MISMATCHES")
print(
    over4_difference
    .fillna(0)
    .ne(0)
    .sum()
)

# --------------------------------------------
# 8. Attendance balance
# under 4h + over 4h should equal total
# --------------------------------------------

attendance_balance = (
    nhs_ae["total_attendances"]
    - (
        nhs_ae["total_under_4h"]
        + nhs_ae["total_over_4h"]
    )
)

print("\nTOTAL ATTENDANCE BALANCE MISMATCHES")
print(
    attendance_balance
    .fillna(0)
    .ne(0)
    .sum()
)

# --------------------------------------------
# 9. Percentage range checks
# --------------------------------------------

percentage_columns = [
    "pct_within_4h_all",
    "pct_within_4h_type1",
    "pct_within_4h_type2",
    "pct_within_4h_type3"
]

print("\nINVALID PERCENTAGE VALUES")

for col in percentage_columns:

    invalid = (
        (nhs_ae[col] < 0)
        | (nhs_ae[col] > 100)
    ).sum()

    print(col, ":", invalid)

# --------------------------------------------
# 10. Emergency admissions reconciliation
# --------------------------------------------

emergency_via_ae_difference = (
    nhs_ae["total_emergency_admissions_via_ae"]
    - (
        nhs_ae["emergency_admissions_type1"]
        + nhs_ae["emergency_admissions_type2"]
        + nhs_ae["emergency_admissions_type3_4"]
    )
)

print("\nEMERGENCY ADMISSIONS VIA A&E MISMATCHES")
print(
    emergency_via_ae_difference
    .fillna(0)
    .ne(0)
    .sum()
)

total_emergency_difference = (
    nhs_ae["total_emergency_admissions"]
    - (
        nhs_ae["total_emergency_admissions_via_ae"]
        + nhs_ae["other_emergency_admissions"]
    )
)

print("\nTOTAL EMERGENCY ADMISSIONS MISMATCHES")
print(
    total_emergency_difference
    .fillna(0)
    .ne(0)
    .sum()
)

# --------------------------------------------
# 11. Provider counts by month
# --------------------------------------------

print("\nUNIQUE PROVIDERS BY MONTH")

print(
    nhs_ae.groupby("reporting_month")
    ["provider_code"]
    .nunique()
)
```

```python
print("Duplicate provider-month records:",
      nhs_ae.duplicated(
          subset=["provider_code", "reporting_month"]
      ).sum())

print("Attendance total mismatches:",
      attendance_difference.fillna(0).ne(0).sum())

print("Under-4-hour mismatches:",
      under4_difference.fillna(0).ne(0).sum())

print("Over-4-hour mismatches:",
      over4_difference.fillna(0).ne(0).sum())

print("Attendance balance mismatches:",
      attendance_balance.fillna(0).ne(0).sum())

print("Emergency admissions via A&E mismatches:",
      emergency_via_ae_difference.fillna(0).ne(0).sum())

print("Total emergency admissions mismatches:",
      total_emergency_difference.fillna(0).ne(0).sum())

print("\nPercentage ranges:")

for col in percentage_columns:
    print(
        col,
        "min =", nhs_ae[col].min(),
        "max =", nhs_ae[col].max()
    )

print("\nRows missing provider name or region:")

print(
    nhs_ae.loc[
        nhs_ae["provider_name"].isna()
        | nhs_ae["region"].isna(),
        [
            "reporting_month",
            "provider_code",
            "region",
            "provider_name"
        ]
    ].to_string(index=False)
)
```

```python
import numpy as np

# ============================================
# FINAL CLEANING AND VALIDATION
# ============================================

# Save the non-provider note separately
quality_notes = nhs_ae[
    nhs_ae["provider_name"].isna()
    | nhs_ae["region"].isna()
].copy()

print("DATA QUALITY NOTES")
print(
    quality_notes[
        ["reporting_month", "provider_code"]
    ].to_string(index=False)
)

# Remove non-provider note rows
nhs_ae_clean = nhs_ae[
    nhs_ae["provider_name"].notna()
    & nhs_ae["region"].notna()
].copy()

nhs_ae_clean = nhs_ae_clean.reset_index(drop=True)

# --------------------------------------------
# Validate percentage fields
# NHS stores percentages as decimal proportions
# e.g. 0.75 = 75%
# --------------------------------------------

percentage_columns = [
    "pct_within_4h_all",
    "pct_within_4h_type1",
    "pct_within_4h_type2",
    "pct_within_4h_type3"
]

print("\nPERCENTAGE VALIDATION")

for col in percentage_columns:
    invalid = (
        (nhs_ae_clean[col] < 0)
        | (nhs_ae_clean[col] > 1)
    ).sum()

    print(
        col,
        "| invalid =", invalid,
        "| min =", round(nhs_ae_clean[col].min(), 4),
        "| max =", round(nhs_ae_clean[col].max(), 4)
    )

# --------------------------------------------
# Validate overall 4-hour performance
# against attendance counts
# --------------------------------------------

calculated_4h_rate = (
    nhs_ae_clean["total_under_4h"]
    / nhs_ae_clean["total_attendances"]
)

valid_rate_rows = (
    nhs_ae_clean["total_attendances"] > 0
) & nhs_ae_clean["pct_within_4h_all"].notna()

rate_difference = (
    calculated_4h_rate[valid_rate_rows]
    - nhs_ae_clean.loc[
        valid_rate_rows,
        "pct_within_4h_all"
    ]
).abs()

print("\n4-HOUR RATE CHECK")
print(
    "Rows checked:",
    valid_rate_rows.sum()
)

print(
    "Maximum absolute difference:",
    rate_difference.max()
)

print(
    "Mismatches > 0.0001:",
    (rate_difference > 0.0001).sum()
)

# --------------------------------------------
# Final row counts
# --------------------------------------------

print("\nFINAL DATASET")
print("Shape:", nhs_ae_clean.shape)

print("\nRows by month:")
print(
    nhs_ae_clean.groupby("reporting_month")
    .size()
)

print(
    "\nUnique providers:",
    nhs_ae_clean["provider_code"].nunique()
)

# --------------------------------------------
# Save final cleaned dataset
# --------------------------------------------

output_file = (
    processed_folder
    / "nhs_ae_provider_apr_aug_2026_clean.csv"
)

nhs_ae_clean.to_csv(
    output_file,
    index=False
)

# Save data-quality note separately
quality_notes.to_csv(
    processed_folder / "nhs_ae_data_quality_notes.csv",
    index=False
)

print("\nSaved cleaned dataset to:")
print(output_file.resolve())
```

```python
# ============================================
# NHS A&E MONTHLY HEADLINE PERFORMANCE
# ============================================

monthly_performance = (
    nhs_ae_clean
    .groupby("reporting_month")
    .agg(
        providers=("provider_code", "nunique"),
        total_attendances=("total_attendances", "sum"),
        attendances_under_4h=("total_under_4h", "sum"),
        attendances_over_4h=("total_over_4h", "sum"),
        emergency_admissions=("total_emergency_admissions", "sum"),
        decision_to_admit_over_4h=("decision_to_admit_over_4h", "sum"),
        decision_to_admit_over_12h=("decision_to_admit_over_12h", "sum")
    )
    .reset_index()
)

# Weighted 4-hour performance
monthly_performance["four_hour_performance_pct"] = (
    100
    * monthly_performance["attendances_under_4h"]
    / monthly_performance["total_attendances"]
)

# Percentage waiting more than 4 hours
monthly_performance["over_4h_pct"] = (
    100
    * monthly_performance["attendances_over_4h"]
    / monthly_performance["total_attendances"]
)

monthly_performance["four_hour_performance_pct"] = (
    monthly_performance["four_hour_performance_pct"]
    .round(2)
)

monthly_performance["over_4h_pct"] = (
    monthly_performance["over_4h_pct"]
    .round(2)
)

print(
    monthly_performance.to_string(index=False)
)
```

```python
# ============================================
# NHS A&E REGIONAL PERFORMANCE
# ============================================

regional_performance = (
    nhs_ae_clean
    .groupby("region")
    .agg(
        providers=("provider_code", "nunique"),
        total_attendances=("total_attendances", "sum"),
        attendances_under_4h=("total_under_4h", "sum"),
        attendances_over_4h=("total_over_4h", "sum"),
        emergency_admissions=("total_emergency_admissions", "sum"),
        decision_to_admit_over_4h=("decision_to_admit_over_4h", "sum"),
        decision_to_admit_over_12h=("decision_to_admit_over_12h", "sum")
    )
    .reset_index()
)

regional_performance["four_hour_performance_pct"] = (
    100
    * regional_performance["attendances_under_4h"]
    / regional_performance["total_attendances"]
)

regional_performance["over_4h_pct"] = (
    100
    * regional_performance["attendances_over_4h"]
    / regional_performance["total_attendances"]
)

regional_performance["four_hour_performance_pct"] = (
    regional_performance["four_hour_performance_pct"]
    .round(2)
)

regional_performance["over_4h_pct"] = (
    regional_performance["over_4h_pct"]
    .round(2)
)

regional_performance = regional_performance.sort_values(
    "four_hour_performance_pct",
    ascending=False
)

print(
    regional_performance.to_string(index=False)
)
```

```python
# ============================================
# TYPE 1 MAJOR A&E PROVIDER BENCHMARKING
# ============================================

provider_performance = (
    nhs_ae_clean
    .groupby(
        ["provider_code", "provider_name", "region"]
    )
    .agg(
        type1_attendances=("type1_attendances", "sum"),
        type1_under_4h=("type1_under_4h", "sum"),
        type1_over_4h=("type1_over_4h", "sum"),
        total_emergency_admissions=(
            "total_emergency_admissions",
            "sum"
        ),
        decision_to_admit_over_12h=(
            "decision_to_admit_over_12h",
            "sum"
        ),
        months_reported=("reporting_month", "nunique")
    )
    .reset_index()
)

# Keep providers with genuine Type 1 Major A&E activity
provider_performance = provider_performance[
    provider_performance["type1_attendances"] > 0
].copy()

# Calculate weighted Type 1 four-hour performance
provider_performance["type1_four_hour_pct"] = (
    100
    * provider_performance["type1_under_4h"]
    / provider_performance["type1_attendances"]
)

provider_performance["type1_four_hour_pct"] = (
    provider_performance["type1_four_hour_pct"]
    .round(2)
)

print("Type 1 providers:",
      len(provider_performance))

print("\nTOP 10 TYPE 1 PROVIDERS")
print(
    provider_performance
    .sort_values(
        "type1_four_hour_pct",
        ascending=False
    )
    [
        [
            "provider_name",
            "region",
            "type1_attendances",
            "type1_four_hour_pct",
            "months_reported"
        ]
    ]
    .head(10)
    .to_string(index=False)
)

print("\nBOTTOM 10 TYPE 1 PROVIDERS")
print(
    provider_performance
    .sort_values(
        "type1_four_hour_pct",
        ascending=True
    )
    [
        [
            "provider_name",
            "region",
            "type1_attendances",
            "type1_four_hour_pct",
            "months_reported"
        ]
    ]
    .head(10)
    .to_string(index=False)
)
```

```python
# ============================================
# MONTHLY A&E DEMAND MIX
# ============================================

monthly_demand = (
    nhs_ae_clean
    .groupby("reporting_month")
    .agg(
        type1_attendances=("type1_attendances", "sum"),
        type2_attendances=("type2_attendances", "sum"),
        type3_attendances=("type3_attendances", "sum"),
        total_attendances=("total_attendances", "sum"),
        emergency_admissions_via_ae=(
            "total_emergency_admissions_via_ae",
            "sum"
        )
    )
    .reset_index()
)

monthly_demand["type1_share_pct"] = (
    100
    * monthly_demand["type1_attendances"]
    / monthly_demand["total_attendances"]
)

monthly_demand["type2_share_pct"] = (
    100
    * monthly_demand["type2_attendances"]
    / monthly_demand["total_attendances"]
)

monthly_demand["type3_share_pct"] = (
    100
    * monthly_demand["type3_attendances"]
    / monthly_demand["total_attendances"]
)

monthly_demand["ae_admission_rate_pct"] = (
    100
    * monthly_demand["emergency_admissions_via_ae"]
    / monthly_demand["total_attendances"]
)

for col in [
    "type1_share_pct",
    "type2_share_pct",
    "type3_share_pct",
    "ae_admission_rate_pct"
]:
    monthly_demand[col] = monthly_demand[col].round(2)

print(monthly_demand.to_string(index=False))

# Save for Power BI
monthly_demand.to_csv(
    processed_folder /
    "nhs_monthly_demand_mix_apr_aug_2026.csv",
    index=False
)
```

```python
# ============================================
# MONTHLY EXECUTIVE KPI TABLE
# ============================================

monthly_kpis = (
    nhs_ae_clean
    .groupby("reporting_month")
    .agg(
        total_attendances=("total_attendances", "sum"),

        type1_attendances=("type1_attendances", "sum"),
        type2_attendances=("type2_attendances", "sum"),
        type3_attendances=("type3_attendances", "sum"),

        attendances_under_4h=("total_under_4h", "sum"),
        attendances_over_4h=("total_over_4h", "sum"),

        emergency_admissions_via_ae=(
            "total_emergency_admissions_via_ae",
            "sum"
        ),

        decision_to_admit_over_4h=(
            "decision_to_admit_over_4h",
            "sum"
        ),

        decision_to_admit_over_12h=(
            "decision_to_admit_over_12h",
            "sum"
        )
    )
    .reset_index()
)

# Four-hour performance
monthly_kpis["four_hour_performance_pct"] = (
    100
    * monthly_kpis["attendances_under_4h"]
    / monthly_kpis["total_attendances"]
)

# Percentage waiting over four hours
monthly_kpis["over_4h_pct"] = (
    100
    * monthly_kpis["attendances_over_4h"]
    / monthly_kpis["total_attendances"]
)

# A&E admission rate
monthly_kpis["ae_admission_rate_pct"] = (
    100
    * monthly_kpis["emergency_admissions_via_ae"]
    / monthly_kpis["total_attendances"]
)

# 12-hour waits per 1,000 A&E admissions
monthly_kpis["twelve_hour_waits_per_1000"] = (
    1000
    * monthly_kpis["decision_to_admit_over_12h"]
    / monthly_kpis["emergency_admissions_via_ae"]
)

# 4-hour decision-to-admit waits per 1,000
monthly_kpis["dta_over_4h_per_1000"] = (
    1000
    * monthly_kpis["decision_to_admit_over_4h"]
    / monthly_kpis["emergency_admissions_via_ae"]
)

# Round percentage/rate columns
rate_columns = [
    "four_hour_performance_pct",
    "over_4h_pct",
    "ae_admission_rate_pct",
    "twelve_hour_waits_per_1000",
    "dta_over_4h_per_1000"
]

monthly_kpis[rate_columns] = (
    monthly_kpis[rate_columns].round(2)
)

print(monthly_kpis.to_string(index=False))

# Save final monthly KPI dataset
monthly_kpis.to_csv(
    processed_folder /
    "nhs_monthly_executive_kpis_apr_aug_2026.csv",
    index=False
)
```
